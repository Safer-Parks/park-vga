This notebook takes the individual per-authority workflows of the previous notebooks and brings them together such that you can provide a list of geojsons and it automatically runs through them and calculates each.

```python
import geopandas as gpd
import os
```

```python
lut = "../example_datasets/LUT_regions_authorities_filenames.geojson"
output_folder = "../workflow_outputs/"
geojson_folder = "../example_datasets/"
```

```python
lut_gdf = gpd.read_file(lut)
```

```python
lut_gdf
```

```python
filenames_list = list(lut_gdf["filename"])
```

```python
filenames_list[0:10]
```

```python
# Extract the folder names from the filenames
# folder_names = filenames_list[0:10] for each entry: .split("_")[0]
folder_names = [filename.split(".")[0] for filename in filenames_list[0:10]]
```

```python
folder_names
```

```python
# check if folders exist in the output folder
# if they do exist, add the existing folders to a new list called existing_folders
# comapre and find duplciated fodlers in both existing_fodler and in the folder_names list
# add these duplciates to a new list called duplicated_folders

existing_folders = os.listdir(output_folder)
duplicated_folders = [folder for folder in folder_names if folder in existing_folders]
```

```python
duplicated_folders
```

```python
# for each folder in duplicated folders,
# open the gejson called {duplicated_folder}.geojson in the geojson folder
# and check the number of features in the geojson file
# if the number of features is greater than 0, print the name of the folder and the number of features in the geojson file
for folder in duplicated_folders:
    geojson_file = os.path.join(geojson_folder, f"{folder}.geojson")
    if os.path.exists(geojson_file):
        gdf = gpd.read_file(geojson_file)
        num_features = len(gdf)
        if num_features > 0:
            print(f"Folder: {folder}, Number of features: {num_features}")
            # comapre the number of features to the number of files in the duplicated folder
            num_files = len(os.listdir(os.path.join(output_folder, folder)))
            print(f"Number of files in duplicated folder: {num_files}")
            print(f"Expected files (if complete): {num_features * 2}")
            if num_files < num_features * 2:
                print(f"Folder {folder} is incomplete. Expected {num_features * 2} files, but found {num_files}.")
            else:
                print(f"Folder {folder} is complete. Found {num_files} files, which matches the expected {num_features * 2}.")
                # if the folder is complete, remove the folder from the duplicated_folders list
                duplicated_folders.remove(folder)
                print(f"Removed {folder} from duplicated_folders list.")


```
