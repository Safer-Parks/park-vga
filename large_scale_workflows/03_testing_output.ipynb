Checking the output of the model runs to ensure they are outputting sensible data.

```python
import os
import random
import geopandas as gpd
```

```python
data_folder = "../workflow_outputs/Blaenau Gwent_pp_or_g_cmb"
```

```python
# select a random file from the folder to test
# want both a foliage and a visibility file to test
files = os.listdir(data_folder)
visibility_files = [f for f in files if "visibility" in f]
foliage_files = [f for f in files if "foliage" in f]
random_visibility_file = random.choice(visibility_files)
random_foliage_file = random.choice(foliage_files)
```

```python
vis = gpd.read_file(os.path.join(data_folder, random_visibility_file))
foliage = gpd.read_file(os.path.join(data_folder, random_foliage_file))
```

```python
vis.plot(column="visibility_pct", cmap="viridis", legend=True)
```
