```python
import requests
import os
from pathlib import Path
```

```python
def download_wales_lidar(output_folder, dtm_url, dsm_url):
    """
    Download Wales LiDAR DTM and DSM mosaics to specified folder.
    
    Parameters:
    -----------
    output_folder : str
        Path where files will be saved
    dtm_url : str
        URL to DTM file
    dsm_url : str
        URL to DSM file
    """
    
    output_path = Path(output_folder)
    output_path.mkdir(parents=True, exist_ok=True)
    print(f"Output folder: {output_path.absolute()}")
    
    urls = {
        'DTM': dtm_url,
        'DSM': dsm_url
    }
    
    for data_type, url in urls.items():
        filename = Path(url).name  # Extract filename from URL
        filepath = output_path / filename
        
        # Check if file already exists
        if filepath.exists():
            print(f"\n{data_type} already exists: {filepath}")
            continue
        
        print(f"\nDownloading {data_type}...")
        print(f"URL: {url}")
        
        try:
            response = requests.get(url, stream=True)
            response.raise_for_status()
            
            total_size = int(response.headers.get('content-length', 0))
            downloaded = 0
            
            with open(filepath, 'wb') as f:
                for chunk in response.iter_content(chunk_size=8192):
                    if chunk:
                        f.write(chunk)
                        downloaded += len(chunk)
                        
                        # Print progress
                        if total_size > 0:
                            percent = (downloaded / total_size) * 100
                            mb_downloaded = downloaded / (1024 * 1024)
                            mb_total = total_size / (1024 * 1024)
                            print(f"\r{data_type}: {percent:.1f}% ({mb_downloaded:.1f}/{mb_total:.1f} MB)", end='')
            
            print(f"\n✓ {data_type} downloaded successfully")
            
        except Exception as e:
            print(f"\n✗ Error downloading {data_type}: {e}")
            if filepath.exists():
                filepath.unlink()
```

```python
# instead of both in one, do a separate download for each, so we can easily re-run just one if needed

def download_single_wales_lidar(output_folder, url, type_name):
    """
    Download Wales LiDAR DTM or DSM mosaics (whichever url is specified) to specified folder.
    
    Parameters:
    -----------
    output_folder : str
        Path where files will be saved
    url : str
        URL to DTM file or DSM file
    type_name : str
        Name of the data type (e.g. "DTM" or "DSM") for filename
    """
    
    output_path = Path(output_folder)
    output_path.mkdir(parents=True, exist_ok=True)
    print(f"Output folder: {output_path.absolute()}")
    
    urls = {
        str(type_name): url
    }
    
    for data_type, url in urls.items():
        filename = Path(url).name  # Extract filename from URL
        filepath = output_path / filename
        
        # Check if file already exists
        if filepath.exists():
            print(f"\n{data_type} already exists: {filepath}")
            continue
        
        print(f"\nDownloading {data_type}...")
        print(f"URL: {url}")
        
        try:
            response = requests.get(url, stream=True)
            response.raise_for_status()
            
            total_size = int(response.headers.get('content-length', 0))
            downloaded = 0
            
            with open(filepath, 'wb') as f:
                for chunk in response.iter_content(chunk_size=8192):
                    if chunk:
                        f.write(chunk)
                        downloaded += len(chunk)
                        
                        # Print progress
                        if total_size > 0:
                            percent = (downloaded / total_size) * 100
                            mb_downloaded = downloaded / (1024 * 1024)
                            mb_total = total_size / (1024 * 1024)
                            print(f"\r{data_type}: {percent:.1f}% ({mb_downloaded:.1f}/{mb_total:.1f} MB)", end='')
            
            print(f"\n✓ {data_type} downloaded successfully")
            
        except Exception as e:
            print(f"\n✗ Error downloading {data_type}: {e}")
            if filepath.exists():
                filepath.unlink()
```
