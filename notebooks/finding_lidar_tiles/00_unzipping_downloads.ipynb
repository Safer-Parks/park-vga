This notebook unzips downloaded lidar tiles:

- Given a file path, it finds all zip arcives within the folder and unzips them
- It will overwrite existing unzipped versions
- It ignores files beginning with `._` (produced by Mac when moving files to external media)

Note: this can be a slow process for large datasets, e.g. the 2022 composite LiDAR dataset (the full DTM dataset takes approx 90 mins, the first pass DSM multiple hours). However, it only needs to be completed once, and the efficient tiff files are not significantly larger when extracted from the zip archive.

This notebook deals with the [2022 composite dataset, that covers 99% of England](https://environment.data.gov.uk/dataset/54167602-36c8-4b2b-80ec-ebb6267b6b1e)

```python
import zipfile
from pathlib import Path
```

```python
def unzip_folders(dir_path):
    """
    Unzip all zip files in a given directory;
    will overwrite existing unzipped versions.
    """
    directory = Path(dir_path)

    # find all zip files in the named loc
    zip_files = [f for f in directory.glob("*.zip") if not f.name.startswith("._")] # filters out dotfiles
    total = len(zip_files)
    print(f"Total found: {total}")
    counter = 0
    for zip_file in zip_files:
        counter += 1   
        with zipfile.ZipFile(zip_file, "r") as zip_ref:
            zip_ref.extractall(directory)
        print(zip_file.name, f"{counter}/{total}")
```

```python
path_dtm = "/Volumes/Extreme SSD/DTM"

unzip_folders(path_dtm)
```

```python
path_dtm = "/Volumes/Extreme SSD/FZ_DSM"

unzip_folders(path_dtm)
```
